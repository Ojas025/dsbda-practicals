```python
text = "Natural language processing is a field of artificial intelligence that focuses on the interaction between computers and humans. It enables machines to understand, interpret, and generate human language."
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
tokens = word_tokenize(text)
print(tokens)
```

```python
pos_tags = pos_tag(tokens)
print(pos_tags)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

```python
ps = PorterStemmer()
stemmed = [ps.stem(word) for word in filtered_tokens]
print(stemmed)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print(lemmatized)
```

```python
documents = [
    text,
    "Machine learning provides systems the ability to automatically learn.",
    "Artificial intelligence and machine learning are closely related fields."
]
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
feature_names = vectorizer.get_feature_names_out()

print("Features:", feature_names)
print("\nTF-IDF Matrix:\n", tfidf_matrix.toarray())
```
