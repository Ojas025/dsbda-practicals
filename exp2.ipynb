```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)

data = {
    "id": range(1, 31),
    "gender": np.random.choice(["Male", "Female"], 30),
    "age": np.random.randint(17, 25, 30),
    "study_hours": np.random.uniform(1, 10, 30),
    "attendance": np.random.uniform(40, 100, 30),
    "marks": np.random.uniform(10, 25, 30),
}

df = pd.DataFrame(data)
```

```python
df['study_hours'] = df['study_hours'].astype(int)
```

```python
df.loc[3, "study_hours"] = np.nan
df.loc[7, "attendance"] = np.nan

df.loc[12, "study_hours"] = 50
```

```python
df['study_hours'] = df['study_hours'].fillna(df['study_hours'].mean())
df['attendance'] = df['attendance'].fillna(df['attendance'].mean())
```

```python
df.isnull().sum()
```

```python
df['gender'] = df['gender'].replace({ "M": "Male", "F": "Female", "1": "Male", "0": "Female" })
```

```python
# Outlier detection

Q1 = df["study_hours"].quantile(0.25)
Q3 = df['study_hours'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['study_hours'] < lower) | (df['study_hours'] > upper)]
outliers
```

```python
df['study_hours'] = np.where(df['study_hours'] > upper, upper, df['study_hours'])
```

```python
df['study_hours'] = (df["study_hours"] - df["study_hours"].min()) / (df['study_hours'].max() - df['study_hours'].min())
```

```python
df['attendance'] = np.log(df['attendance'])
```
