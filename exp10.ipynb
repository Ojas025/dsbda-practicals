```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df.dtypes
```

```python
for col in df.columns:
    if df[col].dtype == 'float64':
        print(f"{col}: numerical")
    else:
        print(f"{col}: categorical")
```

```python
def plot_histogram(feature, df):
    sns.histplot(x=str(feature), hue='species', bins=20, data=df)
    plt.show()
```

```python
for col in df.columns:
    if df[col].dtype == 'float64':
        plot_histogram(col, df)
```

```python
sns.boxplot(data=df.drop('species', axis=1))
plt.show()
```
